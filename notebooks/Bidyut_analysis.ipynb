```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns   
```

```python

# Keep only numeric columns
num_df = df.select_dtypes(include=[np.number]).copy()

target = "REJECTTPH"

if target not in num_df.columns:
    raise ValueError("REJECTTPH not found as numeric column")

# ---- Optional cleaning (recommended) ----
if "TOTALTPH" in num_df.columns:
    num_df = num_df[num_df["TOTALTPH"] > 50]

num_df = num_df.dropna(subset=[target])

# ---- Correlation ----
corr = num_df.corr(method="pearson")[target].drop(target)

# Top 10 strongest relations
top10 = corr.reindex(corr.abs().sort_values(ascending=False).head(10).index)

# ---- Create 1-row heatmap ----
heat_df = pd.DataFrame(top10.values.reshape(1, -1),
                       columns=top10.index,
                       index=[target])

plt.figure(figsize=(14, 3))
sns.heatmap(
    heat_df,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f",
    linewidths=1
)

plt.title("Top 10 Correlations with REJECTTPH")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# ---- Print ranking ----
print("\nTop 10 correlated parameters with REJECTTPH:\n")
print(top10.sort_values(key=lambda x: x.abs(), ascending=False))
```

# 250 Reject TPH Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# Load and prepare data
# -----------------------------
df = pd.read_csv("cement_mill_2024.csv", encoding="latin1")
df["TIME"] = pd.to_datetime(df["TIME"], format="%d-%m-%Y %H:%M")
df = df.sort_values("TIME").reset_index(drop=True)

# -----------------------------
# Parameters (you can tune these)
# -----------------------------
reject_threshold = 240          # reject considered "very high"
window_hours = 6                # size of continuous window to inspect
min_fraction_high = 0.70        # at least 70% of points in window above threshold
min_avg_reject = 230            # average reject in the window should also be high
context_hours_before = 2        # extra plotting context before event
context_hours_after = 2         # extra plotting context after event

# -----------------------------
# Ensure data is hourly and build candidate windows
# -----------------------------
df["time_diff_hr"] = df["TIME"].diff().dt.total_seconds() / 3600

candidate_windows = []

for i in range(len(df) - window_hours + 1):
    sub = df.iloc[i:i + window_hours].copy()

    # Check if this slice is continuous hourly data
    diffs = sub["TIME"].diff().dropna().dt.total_seconds() / 3600
    if not (diffs == 1).all():
        continue

    frac_high = (sub["REJECT TPH"] >= reject_threshold).mean()
    avg_reject = sub["REJECT TPH"].mean()

    if frac_high >= min_fraction_high and avg_reject >= min_avg_reject:
        candidate_windows.append({
            "start": sub["TIME"].iloc[0],
            "end": sub["TIME"].iloc[-1],
            "frac_high": frac_high,
            "avg_reject": avg_reject,
            "max_reject": sub["REJECT TPH"].max()
        })

# -----------------------------
# Merge overlapping candidate windows
# -----------------------------
merged_windows = []

for win in candidate_windows:
    if not merged_windows:
        merged_windows.append(win)
    else:
        last = merged_windows[-1]

        # overlap or touching windows
        if win["start"] <= last["end"] + pd.Timedelta(hours=1):
            last["end"] = max(last["end"], win["end"])
            last["frac_high"] = max(last["frac_high"], win["frac_high"])
            last["avg_reject"] = max(last["avg_reject"], win["avg_reject"])
            last["max_reject"] = max(last["max_reject"], win["max_reject"])
        else:
            merged_windows.append(win)

# -----------------------------
# Show the detected high-reject windows
# -----------------------------
print("Detected windows where Reject TPH stays near 250 for most of the time:\n")

for k, win in enumerate(merged_windows, 1):
    print(
        f"{k}. Start: {win['start']}, End: {win['end']}, "
        f"Avg Reject: {win['avg_reject']:.2f}, "
        f"Max Reject: {win['max_reject']:.2f}, "
        f"Fraction >= {reject_threshold}: {win['frac_high']:.2f}"
    )

# -----------------------------
# Build plotting cases with context
# -----------------------------
cases = []

for k, win in enumerate(merged_windows, 1):
    plot_start = win["start"] - pd.Timedelta(hours=context_hours_before)
    plot_end = win["end"] + pd.Timedelta(hours=context_hours_after)

    cases.append((
        f"High Reject Window {k}",
        str(plot_start),
        str(plot_end)
    ))

print("\nSuggested cases for plotting:\n")
for c in cases:
    print(c)

# -----------------------------
# Plot same style as earlier
# Primary axis: Reject TPH + Total TPH
# Secondary axis: Separator RPM
# -----------------------------
for label, start, end in cases:

    subset = df[(df["TIME"] >= start) & (df["TIME"] <= end)].copy()

    if subset.empty:
        continue

    fig, ax1 = plt.subplots(figsize=(10, 5))

    # Primary axis
    ax1.plot(subset["TIME"], subset["REJECT TPH"],
             color="red", marker='o', label="Reject TPH")
    ax1.plot(subset["TIME"], subset["TOTAL TPH"],
             color="green", marker='s', label="Total TPH")

    ax1.set_xlabel("Time")
    ax1.set_ylabel("Reject / Total TPH")
    ax1.tick_params(axis='y')

    # Secondary axis
    ax2 = ax1.twinx()
    ax2.plot(subset["TIME"], subset["CM SEP DRIVE Speed (RPM)"],
             color="blue", marker='^', label="Separator RPM")
    ax2.set_ylabel("Separator RPM", color="blue")
    ax2.tick_params(axis='y', colors='blue')

    # Combine legends
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc="best")

    plt.title(f"{label}: Reject vs Feed vs Separator RPM")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```
